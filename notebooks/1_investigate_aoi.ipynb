# Investigate DIST-S1 artifacts over an AOI

Given a lon/lat bbox, get the DIST-S1 alert layers over it and the RTC-S1 inputs used by
the most recent products, with direct download urls for every layer.

```python
import geopandas as gpd

from dist_s1_prod_check import get_products_over_aoi, get_rtc_inputs_for_products
```

```python
BBOX = (-120.0, 34.0, -118.0, 36.0)
START = '2026-02-01'
STOP = '2026-03-01'
N_RECENT = 5
```

## DIST-S1 products over the AOI

Queries CMR directly; pass `df_dist=` with a local geoparquet table to skip the network.

```python
df_products = get_products_over_aoi(BBOX, START, STOP)
print(f'{len(df_products)} products over {df_products.mgrs_tile_id.nunique()} tiles')
df_products[['opera_id', 'mgrs_tile_id', 'acq_time', 'alert_url']].tail()
```

```python
df_products.explore(column='mgrs_tile_id', legend=False)
```

## RTC-S1 inputs for the most recent products

```python
df_recent = df_products.sort_values('acq_time').groupby('mgrs_tile_id').tail(1).tail(N_RECENT)
df_rtc_inputs = get_rtc_inputs_for_products(df_recent)
df_rtc_inputs.groupby(['dist_s1_opera_id', 'input_category']).size()
```
